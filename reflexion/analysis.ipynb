```python
"""
Which games were hardest to save? 
Plotting how many instances per game were left aborted or lost after three retries,
for all models.
"""

import pandas as pd
outcomes = r"C:\Users\white\Desktop\MSc_thesis\baseline_analysis\outcomes.csv"

outcomes_df = pd.read_csv(outcomes)

outcomes_df = outcomes_df[outcomes_df.group =="games"]

rows = []
for key, sub in outcomes_df.groupby(["game"]):
    n = len(sub)
    never = len(sub[sub.outcome != "Success"])
    row = dict(zip(["game", "experiment"], key if isinstance(key, tuple) else (key,)))
    row["n"] = n
    row["never_solved"] = never
    row["share"] = 100 * never / n
    rows.append(row)
    pd.DataFrame(rows).to_csv(r"C:\Users\white\Desktop\MSc_thesis\reflexion\selected_episodes\result_baseline.csv")
```
